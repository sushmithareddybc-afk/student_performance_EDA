```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully!")
```

```python
url = "https://raw.githubusercontent.com/MakroMB/student-performance-analysis/main/data/student-mat.csv"

df = pd.read_csv(url, sep=";")

print("Dataset loaded successfully!")
```

```python
df.head()
```

```python
df.shape
df.columns
df.info()
df.describe()
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nTotal Missing Values:")
print(df.isnull().sum().sum())
```

```python
print("Duplicate Records:")
print(df.duplicated().sum())
```

```python
df_clean = df.drop_duplicates().copy()

print("Original Dataset:", df.shape)
print("Cleaned Dataset:", df_clean.shape)
```

```python
df_clean.info()

df_clean.head()
```

```python
print("Average First Grade:", df_clean['G1'].mean())

print("Average Second Grade:", df_clean['G2'].mean())

print("Average Final Grade:", df_clean['G3'].mean())
```

```python
grades = ['G1', 'G2', 'G3']

averages = [
    df_clean['G1'].mean(),
    df_clean['G2'].mean(),
    df_clean['G3'].mean()
]

plt.figure(figsize=(8,5))
plt.bar(grades, averages)

plt.title('Average Student Grades')
plt.xlabel('Grade Period')
plt.ylabel('Average Marks')

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df_clean['G3'], bins=10)

plt.title('Distribution of Final Grades')
plt.xlabel('Final Grade')
plt.ylabel('Number of Students')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df_clean,
    x='studytime',
    y='G3'
)

plt.title('Study Time vs Final Grades')
plt.xlabel('Study Time')
plt.ylabel('Final Grade')

plt.show()
```

```python
plt.figure(figsize=(12,8))

sns.heatmap(
    df_clean.select_dtypes(include='number').corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Student Performance Correlation Heatmap')
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df_clean['G3'], bins=10)

plt.title('Distribution of Final Grades')
plt.xlabel('Final Grade')
plt.ylabel('Number of Students')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df_clean,
    x='studytime',
    y='G3'
)

plt.title('Study Time vs Final Grades')
plt.xlabel('Study Time')
plt.ylabel('Final Grade')

plt.show()
```

```python
df_clean.to_csv('Student_Cleaned.csv', index=False)

print("Cleaned dataset saved successfully!")
```
